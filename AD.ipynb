```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.model_selection import train_test_split
import xgboost as xgb
import catboost as cb
from flaml import AutoML
from sklearn.metrics import accuracy_score, f1_score, roc_curve, auc, precision_recall_curve, confusion_matrix

def prepare_data(normal_df, abnormal_df, fn_df, useful_cols, test_size=0.2, random_seed=80):
   
    normal_df = normal_df[useful_cols]
    abnormal_df = abnormal_df[useful_cols]
    fn_df = fn_df[useful_cols]

 
    normal_df["Label"] = "normal"
    abnormal_df["Label"] = "abnormal"
    fn_df["Label"] = "fn"

 
    data = pd.concat([normal_df, abnormal_df, fn_df], axis=0).reset_index(drop=True)


    X = data.drop(columns=["Label"])
    y = data["Label"]


    train_x, test_x, train_y, test_y = train_test_split(
        X, y, test_size=test_size, random_state=random_seed, stratify=y
    )


    train_y = train_y.map({"normal": 0, "abnormal": 1, "fn": 1})
    test_y = test_y.map({"normal": 0, "abnormal": 1, "fn": 1})  

    
    print("\n🔹 Train Label Distribution:")
    print(train_y.value_counts())  
    print("\n🔹 Test Label Distribution:")
    print(test_y.value_counts()) 

    return train_x, test_x, train_y, test_y


def get_best_threshold(y_true, y_probs):
    """Precision-Recall Curve에서 F1-score가 가장 높은 threshold 선택"""
    precisions, recalls, thresholds = precision_recall_curve(y_true, y_probs)
    f1_scores = (2 * precisions * recalls) / (precisions + recalls + 1e-9)  
    best_thr = thresholds[np.argmax(f1_scores)] 
    return best_thr

def run_automl(X_train, X_test, y_train, y_test):
    automl = AutoML()
    automl.fit(X_train, y_train, task="classification", max_iter=800) 
    
    joblib.dump(automl, "best_automl_model.pkl")
    
    with open("best_automl_hyperparams.txt", "w") as f:
        f.write(str(automl.best_config))
    
    automl_probs = automl.predict_proba(X_test)[:, 1]  
    
 
    best_thr = get_best_threshold(y_test, automl_probs)
    print(f"🔹 Best Threshold Found: {best_thr:.4f}")


    automl_preds = (automl_probs >= best_thr).astype(int)

    results = {
        'AutoML Accuracy': accuracy_score(y_test, automl_preds),
        'AutoML F1 Score': f1_score(y_test, automl_preds)
    }
    

    plt.figure(figsize=(6, 5))
    sns.heatmap(confusion_matrix(y_test, automl_preds), annot=True, fmt='d', cmap='Purples')
    plt.title("AutoML Confusion Matrix")
    plt.show()

    fpr_auto, tpr_auto, _ = roc_curve(y_test, automl_probs)
    plt.figure(figsize=(8, 6))
    plt.plot(fpr_auto, tpr_auto, label=f'AutoML (AUC = {auc(fpr_auto, tpr_auto):.4f})', color='purple')
    plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("ROC Curve - AutoML")
    plt.legend()
    plt.show()
    

    precision_auto, recall_auto, _ = precision_recall_curve(y_test, automl_probs)
    plt.figure(figsize=(8, 6))
    plt.plot(recall_auto, precision_auto, label=f'AutoML (AUPRC = {auc(recall_auto, precision_auto):.4f})', color='purple')
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("Precision-Recall Curve - AutoML")
    plt.legend()
    plt.show()
    
    return results


```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
import xgboost as xgb
import catboost as cb
from sklearn.metrics import accuracy_score, f1_score, roc_curve, auc, precision_recall_curve, confusion_matrix

# 데이터 준비
def prepare_data(normal_df, abnormal_df, fn_df, test_size=0.15, random_seed=42):
    normal_df = normal_df.iloc[:, 1:]
    abnormal_df = abnormal_df.iloc[:, 1:]
    fn_df = fn_df.iloc[:, 1:]
    
    normal_df['Label'] = 0
    abnormal_df['Label'] = 1
    fn_df['Label'] = 1
    
    data = pd.concat([normal_df, abnormal_df, fn_df], axis=0).sample(frac=1, random_state=random_seed).reset_index(drop=True)
    X = data.drop(columns=['Label'])
    y = data['Label']
    
    return train_test_split(X, y, test_size=test_size, random_state=random_seed)

# 학습 및 평가 함수
def train_and_evaluate(X_train, X_test, y_train, y_test):
    # XGBoost 모델 학습
    xgb_model = xgb.XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42)
    xgb_model.fit(X_train, y_train)
    xgb_preds = xgb_model.predict(X_test)
    xgb_probs = xgb_model.predict_proba(X_test)[:, 1]
    
    # CatBoost 모델 학습
    cat_model = cb.CatBoostClassifier(verbose=0, random_seed=42)
    cat_model.fit(X_train, y_train)
    cat_preds = cat_model.predict(X_test)
    cat_probs = cat_model.predict_proba(X_test)[:, 1]
    
    # 평가 지표 계산
    results = {
        'XGBoost Accuracy': accuracy_score(y_test, xgb_preds),
        'XGBoost F1 Score': f1_score(y_test, xgb_preds),
        'CatBoost Accuracy': accuracy_score(y_test, cat_preds),
        'CatBoost F1 Score': f1_score(y_test, cat_preds)
    }
    
    # Confusion Matrix 시각화
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    sns.heatmap(confusion_matrix(y_test, xgb_preds), annot=True, fmt='d', cmap='Blues', ax=axes[0])
    axes[0].set_title("XGBoost Confusion Matrix")
    sns.heatmap(confusion_matrix(y_test, cat_preds), annot=True, fmt='d', cmap='Reds', ax=axes[1])
    axes[1].set_title("CatBoost Confusion Matrix")
    plt.show()
    
    # ROC Curve
    fpr_xgb, tpr_xgb, _ = roc_curve(y_test, xgb_probs)
    fpr_cat, tpr_cat, _ = roc_curve(y_test, cat_probs)
    plt.figure(figsize=(8, 6))
    plt.plot(fpr_xgb, tpr_xgb, label=f'XGBoost (AUC = {auc(fpr_xgb, tpr_xgb):.5f})', color='blue')
    plt.plot(fpr_cat, tpr_cat, label=f'CatBoost (AUC = {auc(fpr_cat, tpr_cat):.5f})', color='red')
    plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("ROC Curve")
    plt.legend()
    plt.show()
    
    # Precision-Recall Curve
    precision_xgb, recall_xgb, _ = precision_recall_curve(y_test, xgb_probs)
    precision_cat, recall_cat, _ = precision_recall_curve(y_test, cat_probs)
    plt.figure(figsize=(8, 6))
    plt.plot(recall_xgb, precision_xgb, label=f'XGBoost (AUPRC = {auc(recall_xgb, precision_xgb):.5f})', color='blue')
    plt.plot(recall_cat, precision_cat, label=f'CatBoost (AUPRC = {auc(recall_cat, precision_cat):.5f})', color='red')
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("Precision-Recall Curve")
    plt.legend()
    plt.show()
    
    return results
```
