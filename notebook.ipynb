### Normal vs Abnormal+Fn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from scipy.stats import ttest_ind
from sklearn.model_selection import train_test_split
import xgboost as xgb
import catboost as cb
from flaml import AutoML
from sklearn.metrics import accuracy_score, f1_score, roc_curve, auc, precision_recall_curve, confusion_matrix



def prepare_data(normal_df, abnormal_df, fn_df, test_size=0.1, random_seed=42):
    normal_df = normal_df.iloc[:, 1:]
    abnormal_df = pd.concat([abnormal_df.iloc[:, 1:], fn_df.iloc[:, 1:]], axis=0)  # FN을 Abnormal로 포함
    
    normal_df['Label'] = 0
    abnormal_df['Label'] = 1
    
    data = pd.concat([normal_df, abnormal_df], axis=0).sample(frac=1, random_state=random_seed).reset_index(drop=True)
    X = data.drop(columns=['Label'])
    y = data['Label']
    
    return train_test_split(X, y, test_size=test_size, random_state=random_seed)


def run_ttest_and_plot(normal_df, abnormal_df):
    feature_columns = normal_df.columns[1:]  # 첫 번째 열 제외
    ttest_results = []
    
    for feature in feature_columns:
        t_stat, p_value = ttest_ind(normal_df[feature], abnormal_df[feature], equal_var=False)
        
        ttest_results.append({
            'Feature': feature,
            'T-Statistic (Normal vs Abnormal)': t_stat,
            'P-Value (Normal vs Abnormal)': p_value
        })
    
    ttest_df = pd.DataFrame(ttest_results)
    
    # P-value 시각화
    plt.figure(figsize=(12, 6))
    sns.heatmap(ttest_df.set_index('Feature').iloc[:, 1:], annot=True, fmt='.2g', cmap='coolwarm', linewidths=0.5)
    plt.title("T-test P-Values (Normal vs Abnormal including FN)")
    plt.show()
    
    return ttest_df


X_train, X_test, y_train, y_test = prepare_data(normal, abnormal, Fn)
ttest_results = run_ttest_and_plot(normal, pd.concat([abnormal, Fn], axis=0))

print(ttest_results)
```

### (8:2)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.model_selection import train_test_split
import xgboost as xgb
import catboost as cb
from flaml import AutoML
from sklearn.metrics import accuracy_score, f1_score, roc_curve, auc, precision_recall_curve, confusion_matrix

def prepare_data(normal_df, abnormal_df, fn_df, useful_cols, test_size=0.2, random_seed=80):
   
    normal_df = normal_df[useful_cols]
    abnormal_df = abnormal_df[useful_cols]
    fn_df = fn_df[useful_cols]

 
    normal_df["Label"] = "normal"
    abnormal_df["Label"] = "abnormal"
    fn_df["Label"] = "fn"

 
    data = pd.concat([normal_df, abnormal_df, fn_df], axis=0).reset_index(drop=True)


    X = data.drop(columns=["Label"])
    y = data["Label"]


    train_x, test_x, train_y, test_y = train_test_split(
        X, y, test_size=test_size, random_state=random_seed, stratify=y
    )


    train_y = train_y.map({"normal": 0, "abnormal": 1, "fn": 1})
    test_y = test_y.map({"normal": 0, "abnormal": 1, "fn": 1})  

    
    print("\n🔹 Train Label Distribution:")
    print(train_y.value_counts())  
    print("\n🔹 Test Label Distribution:")
    print(test_y.value_counts()) 

    return train_x, test_x, train_y, test_y


def get_best_threshold(y_true, y_probs):
    """Precision-Recall Curve에서 F1-score가 가장 높은 threshold 선택"""
    precisions, recalls, thresholds = precision_recall_curve(y_true, y_probs)
    f1_scores = (2 * precisions * recalls) / (precisions + recalls + 1e-9)  
    best_thr = thresholds[np.argmax(f1_scores)] 
    return best_thr

def run_automl(X_train, X_test, y_train, y_test):
    automl = AutoML()
    automl.fit(X_train, y_train, task="classification", max_iter=800) 
    
    joblib.dump(automl, "best_automl_model.pkl")
    
    with open("best_automl_hyperparams.txt", "w") as f:
        f.write(str(automl.best_config))
    
    automl_probs = automl.predict_proba(X_test)[:, 1]  
    
 
    best_thr = get_best_threshold(y_test, automl_probs)
    print(f"🔹 Best Threshold Found: {best_thr:.4f}")


    automl_preds = (automl_probs >= best_thr).astype(int)

    results = {
        'AutoML Accuracy': accuracy_score(y_test, automl_preds),
        'AutoML F1 Score': f1_score(y_test, automl_preds)
    }
    

    plt.figure(figsize=(6, 5))
    sns.heatmap(confusion_matrix(y_test, automl_preds), annot=True, fmt='d', cmap='Purples')
    plt.title("AutoML Confusion Matrix")
    plt.show()

    fpr_auto, tpr_auto, _ = roc_curve(y_test, automl_probs)
    plt.figure(figsize=(8, 6))
    plt.plot(fpr_auto, tpr_auto, label=f'AutoML (AUC = {auc(fpr_auto, tpr_auto):.4f})', color='purple')
    plt.plot([0, 1], [0, 1], linestyle='--', color='gray')
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("ROC Curve - AutoML")
    plt.legend()
    plt.show()
    

    precision_auto, recall_auto, _ = precision_recall_curve(y_test, automl_probs)
    plt.figure(figsize=(8, 6))
    plt.plot(recall_auto, precision_auto, label=f'AutoML (AUPRC = {auc(recall_auto, precision_auto):.4f})', color='purple')
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("Precision-Recall Curve - AutoML")
    plt.legend()
    plt.show()
    
    return results


```

```python

X_train, X_test, y_train, y_test = prepare_data(normal, abnormal, Fn, useful_cols)
auto_ml_results = run_automl(X_train, X_test, y_train, y_test)

print(auto_ml_results)
```
